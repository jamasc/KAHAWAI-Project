```python
# this code adds a label column to the datasheet
# in this case the label depends on if the values of a certain column are in the top x%

import pandas as pd

percentile = 0.95
file_path = 'pukele_datasheet.xlsx'
df = pd.read_excel(file_path)

threshold = df['discharge_cfs'].quantile(percentile)

df['top_5_percent'] = df['discharge_cfs'].apply(lambda x: 1 if x >= threshold else 0)

# Step 4: Save the updated DataFrame to a new Excel file
output_file = 'updated_file.xlsx'
df.to_excel(output_file, index=False)

print(f"New column 'top_5_percent' added. Updated file saved as '{output_file}'.")
```
