```python
# this code provides stratified datasheets for the DAR dataset (same amount of datapoints per label and cam)

import pandas as pd

# ===== USER SETTINGS =====
samples_per_label = 41                  # number of samples per label
random_seed = 42                         # for reproducibility
labels_to_remove = [5, 6]
# ==========================

for i in range(7):
    input_excel_path = f'DAR_cam_100{i}_datasheet.xlsx'
    output_excel_path = f'DAR_100{i}_stratified.xlsx'

    print(f"===== {input_excel_path} =====")
    
    # Load Excel file
    df = pd.read_excel(input_excel_path)
    
    # Basic checks
    if "path" not in df.columns or "label" not in df.columns:
        raise ValueError("Excel file must contain 'path' and 'label' columns.")
    
    print(f"Total samples: {len(df)}")

    # 🔥 Remove unwanted labels
    df = df[~df["label"].isin(labels_to_remove)].reset_index(drop=True)
    
    print(f"Total samples after removing {labels_to_remove}: {len(df)}")
    
    # 1️⃣ Count label instances
    label_counts = df["label"].value_counts().sort_index()
    print("\nOriginal label distribution:")
    print(label_counts)
    
    # 2️⃣ Create balanced dataset (random sampling per label)
    balanced_df_list = []
    
    for label, group in df.groupby("label"):
        if len(group) < samples_per_label:
            print(f"Warning: Label {label} has only {len(group)} samples (< {samples_per_label})")
            sampled_group = group.sample(n=len(group), random_state=random_seed)
        else:
            sampled_group = group.sample(n=samples_per_label, random_state=random_seed)
        
        balanced_df_list.append(sampled_group)
    
    balanced_df = pd.concat(balanced_df_list).reset_index(drop=True)
    
    # Shuffle full dataset
    balanced_df = balanced_df.sample(frac=1, random_state=random_seed).reset_index(drop=True)
    
    print("\nBalanced label distribution:")
    print(balanced_df["label"].value_counts().sort_index())
    
    # Save new Excel file
    balanced_df.to_excel(output_excel_path, index=False)
    
    print(f"\nBalanced dataset saved to: {output_excel_path}")
    print(f"Final dataset size: {len(balanced_df)}")
```
