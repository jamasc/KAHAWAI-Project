```python
# this code is to obtain balanced DAR data fro training (same amount of datapoints for each label)

import pandas as pd
import numpy as np

# ---------------------------------------------------
# CONFIG
input_file = "camera_files/DAR_cam_1006_datasheet.xlsx"        # your original dataset
output_file = "camera_files/DAR_1006_balanced.xlsx"    # output dataset
label_col = "label"              # column holding labels
# ---------------------------------------------------

# Load data
df = pd.read_excel(input_file)

# Remove labels 5 and 6
df = df[~df[label_col].isin([5, 6])]

# Count how many rows have label 4
count_label4 = (df[label_col] == 4).sum()

print("Number of datapoints with label 4:", count_label4)

# List of labels to include (except 4)
candidate_labels = [1, 2, 3]

balanced_parts = []

# Always include all label 4 datapoints
balanced_parts.append(df[df[label_col] == 4])

# Sample same number from each other label (if they exist)
for lbl in candidate_labels:
    subset = df[df[label_col] == lbl]
    if len(subset) == 0:
        print(f"Label {lbl} does not exist, skipping.")
        continue

    # If subset has fewer than needed, sample with replacement
    replace = len(subset) < count_label4
    sampled = subset.sample(n=count_label4, replace=replace, random_state=42)
    balanced_parts.append(sampled)

# Combine into new balanced dataset
balanced_df = pd.concat(balanced_parts, ignore_index=True)

# Save result
balanced_df.to_excel(output_file, index=False)

print("Balanced dataset saved to:", output_file)
balanced_df.head()
```
