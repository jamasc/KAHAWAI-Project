```python
# this code is to write out the sheet names from the data sheets and the image folder names
# to make sure it all adds up

# INPUT: need the locations of the xlsx files with the metadata

### DATA SHEETS ###
from openpyxl import load_workbook

koa_xlsx_files = [
    "../data/2019_Data/Stream Camera Data.xlsx",
    "../data/2020_Data/2020_Stream_Cam_Data.xlsx",
    "../data/2020_Data/2020_Stream_Cam_Data_2.xlsx",
    "../data/2020_2021_Data/2021_Stream_Cam_Data_2.xlsx"
]
hd_xlsx_files = [
    "D:/Stream Camera Project/2019_StreamCams/2019_Data/Stream Camera Data.xlsx",
    "D:/Stream Camera Project/2020_StreamCams/Data/2020_Stream_Cam_Data.xlsx",
    "D:/Stream Camera Project/2020_StreamCams/Data/2020_Stream_Cam_Data_2.xlsx",
    "D:/Stream Camera Project/2021_2022_StreamCams/Data/2021_Stream_Cam_Data_2.xlsx"
]

# creates a list of sheets from the excel files
def createSheetList(p=True):
    if p: print("Creating sheet list")
    xlsx_files = hd_xlsx_files
    sheet_list = []
    count = 0
    for file_path in xlsx_files:
        workbook = load_workbook(file_path, read_only=True)
        if p: print(f"File: {file_path}")
        for sheet_name in workbook.sheetnames:

            img_count = 0
            sheet = workbook[sheet_name]
            h_row = sheet[2]
            target_col = None
            for idx, cell in enumerate(h_row, start=1):
                if str(cell.value).strip().lower() == 'picture number camera output':
                    target_col = idx
                    break
            if target_col is not None:
                for row in sheet.iter_rows(min_row=3, min_col=target_col, max_col=target_col):
                    if row[0].value is not None:
                        img_count += 1
            
            sheet_list.append([sheet_name, file_path, img_count])  
            count += 1
            if p: print(f"  {count}: {sheet_name}")
        if p: print("-" * 40)
    print(f"{count} sheets found")
    return sheet_list  

print("done")
```

```python
sheet_list = createSheetList(False)

for sheet in sheet_list:
    print(sheet)
```

```python
### IMAGE FOLDERS ###
import os
from pathlib import Path

# creates a list of folders containing images
def createFolderList(p=True):
    root = Path("E:/backup/Stream Camera Project")  # Change this to your folder
    image_extensions = {'.jpg', '.jpeg', '.png', '.bmp', '.gif', '.tiff', '.webp'}
    folder_list = []

    subdirs_with_brackets = [d for d in root.rglob('*') if d.is_dir() and '(' in d.name and ')' in d.name]
    print(len(subdirs_with_brackets))
    for subdir in subdirs_with_brackets:
        # Count images directly in the subdir
        images_in_subdir = sum(1 for f in subdir.iterdir()
                               if f.is_file() and f.suffix.lower() in image_extensions)
        
        # Find MEDIA dirs inside the subdir
        media_dirs = [child for child in subdir.iterdir() if child.is_dir() and 'MEDIA' in child.name.upper()]
        media_dirs_count = len(media_dirs) 
        
        # Count images inside each MEDIA dir (non-recursive)
        images_in_media = 0
        for media_dir in media_dirs:
            images_in_media += sum(1 for f in media_dir.iterdir()
                                  if f.is_file() and f.suffix.lower() in image_extensions)
        
        total_images = images_in_subdir + images_in_media

        folder_list.append([str(subdir.resolve()), media_dirs_count, total_images])

    return folder_list
```

```python
folder_list = createFolderList()

for folder in folder_list:
    print(folder)
```

```python
# creates an inventory of all the sheets and folders

import re
import pandas as pd

def str2id(s):
    # Extract id1 and id2 inside parentheses
    ids_match = re.search(r'\(([^,]+),\s*([^)]+)\)', s)
    id1 = ids_match.group(1).strip() if ids_match else None
    id2 = ids_match.group(2).strip() if ids_match else None

    # Extract d after deploy/deployment/recovery
    d_match = re.search(r'\b(?:deploy|deployment|recovery)\b\s+(\S+)', s, re.IGNORECASE)
    d = d_match.group(1).strip() if d_match else None

    return id1, id2, d
    
def createDataInventory(sheet_list, folder_list):
    ids = {}
    inventory = []
    for sheet in sheet_list:
        sheet_id = str2id(sheet[0])
        if sheet_id not in ids:
            ids[sheet_id] = [0, 0]
        ids[sheet_id][0] += 1
        inventory.append([sheet_id, sheet[2], sheet[1], sheet[0]])
        
    for folder in folder_list:
        folder_id = str2id(folder[0])
        if folder_id not in ids:
            ids[folder_id] = [0, 0]
        ids[folder_id][1] += 1
        inventory.append([folder_id, folder[2], folder[0], ''])

    for key, value in ids.items():
        print(f"{key} : {value}")

    columns = ['ID', '# imgs', 'path', 'name']
    inventory_sorted = sorted(inventory, key=lambda x: str(x[0]) if x[0] is not None else '')
    df = pd.DataFrame(inventory_sorted, columns=columns)
    df.to_excel('data_inventory2.xlsx', index=False)

    print('done')
    
    return

createDataInventory(sheet_list, folder_list)
```
