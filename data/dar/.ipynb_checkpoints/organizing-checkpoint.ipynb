```python
# FOLDER NAMES

import os

# Folder to scan
root_folder = "../images"  # Change this to your folder

# Allowed image extensions
image_extensions = {".jpg", ".jpeg", ".png", ".bmp", ".gif", ".tiff", ".webp"}

for dirpath, dirnames, filenames in os.walk(root_folder):
    # Skip the root folder itself if you only want subdirs
    if dirpath == root_folder:
        continue

    # Check if there are any subdirectories inside this one
    if dirnames:
        continue  # Skip if it contains subdirectories

    # Check if all files are images (or it's empty)
    if all(os.path.splitext(f)[1].lower() in image_extensions for f in filenames):
        print(dirpath)
```

```python
# creates a nice list

import os
import re
import pandas as pd
from openpyxl import load_workbook

# --- SETTINGS ---

# Excel files to scan
excel_files = [
    "../data/2019_Data/Stream Camera Data.xlsx",
    "../data/2020_Data/2020_Stream_Cam_Data.xlsx",
    "../data/2020_Data/2020_Stream_Cam_Data_2.xlsx",
    "../data/2020_2021_Data/2021_Stream_Cam_Data_2.xlsx"
]

# Image folder root directory
image_root = "../images"

# Pattern to extract IDs inside ()
id_pattern = re.compile(r"\(([^()]+),\s*([^()]+)\)")
deployment_pattern = re.compile(r"\b(?:deploy(?:ment)?|recovery)\s*(\d+)", re.IGNORECASE)

# --- DATA COLLECTION ---

records = {}

# 1) Scan Excel Files
for file_path in excel_files:
    workbook = load_workbook(file_path, read_only=True)
    for sheet_name in workbook.sheetnames:
        id_match = id_pattern.search(sheet_name)
        dep_match = deployment_pattern.search(sheet_name)
        if id_match and dep_match:
            id1, id2 = id_match.group(1), id_match.group(2)
            deployment = dep_match.group(1)
            key = (id1, id2, deployment)
            records.setdefault(key, {"excel_sources": [], "image_folders": []})
            relative_path = f"data/{os.path.basename(file_path)} :: {sheet_name}"
            records[key]["excel_sources"].append(relative_path)

# 2) Scan Image Folders
for dirpath, _, filenames in os.walk(image_root):
    # Optionally skip if no images, or leave this out if you want all matching folders
    if not any(f.lower().endswith((".jpg", ".jpeg", ".png", ".bmp", ".gif", ".tiff", ".webp")) for f in filenames):
        continue  

    id_match = id_pattern.search(dirpath)
    dep_match = deployment_pattern.search(dirpath)
    if id_match and dep_match:
        id1, id2 = id_match.group(1), id_match.group(2)
        deployment = dep_match.group(1)
        key = (id1, id2, deployment)
        records.setdefault(key, {"excel_sources": [], "image_folders": []})
        records[key]["image_folders"].append(dirpath)


# --- WRITE TO EXCEL ---

output_rows = []

for (id1, id2, deployment), data in records.items():
    # Max number of rows we need for this ID/deployment combo
    num_rows = max(len(data["excel_sources"]), len(data["image_folders"]), 1)

    for i in range(num_rows):
        row = {
            "ID1": id1 if i == 0 else "",
            "ID2": id2 if i == 0 else "",
            "Deployment": deployment if i == 0 else "",
            "Excel Source": data["excel_sources"][i] if i < len(data["excel_sources"]) else "",
            "Image Folder": data["image_folders"][i] if i < len(data["image_folders"]) else "",
        }
        output_rows.append(row)

df = pd.DataFrame(output_rows)
df.to_excel("Data_Inventory.xlsx", index=False)
```
