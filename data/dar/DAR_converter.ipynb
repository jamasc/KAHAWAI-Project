```python
# this little script turns the image_inventory sheet of the DAR dataset into several sheets for each camera and their labels

import pandas as pd
import os

excel_file = "image_inventory.xlsx"
output_dir = "camera_files"
os.makedirs(output_dir, exist_ok=True)

# Read all sheets; returns a dictionary: {sheet_name: dataframe}
all_sheets = pd.read_excel(excel_file, sheet_name=None)
all_sheets.keys()

# Dictionary to accumulate rows per Camera_ID
camera_dict = {}

for label, df in all_sheets.items():
    # Ensure required columns exist
    if "Image_Path" not in df.columns or "Camera_ID" not in df.columns:
        print(f"Skipping sheet '{label}': missing required columns.")
        continue
    
    # Add label column
    df = df.copy()
    df["Label"] = int(label.split("_")[-1])
    
    # Group by Camera_ID
    for _, row in df.iterrows():
        cam_id = row["Camera_ID"]
        if cam_id not in camera_dict:
            camera_dict[cam_id] = []
        camera_dict[cam_id].append({"path": row["Image_Path"], "label": row["Label"]})

for cam_id, rows in camera_dict.items():
    cam_df = pd.DataFrame(rows)[["path", "label"]]
    
    # Output file name
    outfile = os.path.join(output_dir, f"DAR_cam_{cam_id}_datasheet.xlsx")
    
    # Save the file
    cam_df.to_excel(outfile, index=False)

print("Finished creating files!")
```
