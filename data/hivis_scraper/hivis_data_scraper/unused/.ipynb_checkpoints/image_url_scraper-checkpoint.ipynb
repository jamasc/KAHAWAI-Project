```python
# Image scraper

# TODO
# [ ] try downloading a single image from a given url and see if it works
# [ ] make this script work to get image url
# then:
# [ ] make big loopy script that adds the image url to the xlsx sheet
# [ ] IF ALL WORKS: maker another script to actually download images from the excel sheets

from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time
import requests

# Start browser (make sure chromedriver is installed & matches Chrome version)
driver = webdriver.Chrome()
driver.get("https://apps.usgs.gov/hivis/camera/NC_Little_Sugar_Creek_at_Medical_Center_Dr_at_Charlotte")

wait = WebDriverWait(driver, 15)

# 1. Open the date selector
date_button = wait.until(EC.element_to_be_clickable((By.CSS_SELECTOR, "span[aria-label='Open Date Selection Panel'] button")))
date_button.click()

# 2. TODO: navigate the calendar widget and select your date/time
#    (this part depends on how their datepicker is structured — you’ll need to inspect with F12)

# For now, pause so you can pick manually and test
time.sleep(10)

# 3. Wait for the displayed image to appear
img = wait.until(EC.presence_of_element_located((By.CSS_SELECTOR, "img[alt='Displayed']")))
img_url = img.get_attribute("src")
print("Image URL:", img_url)

driver.quit()
```

```python
# 4. Download it
r = requests.get(img_url)
with open(img_url.split("/")[-1], "wb") as f:
    f.write(r.content)
print("Saved", img_url.split("/")[-1])
```
