```python
# Image url builder
# this code builds the correct url for getting HIVIS images

import requests
from datetime import datetime, timedelta

SITE = "NC_Little_Sugar_Creek_at_Medical_Center_Dr_at_Charlotte"

def build_url(dt):
    # Try a few second offsets around the desired minute mark
    for sec in range(0, 60):
        ts = dt.replace(second=sec)
        ts_str = ts.strftime("%Y-%m-%dT%H-%M-%SZ")
        url = (
            f"https://apps.usgs.gov/hivis/camera/{SITE}"
            f"#&gid=hivis&pid={SITE}___{ts_str}.jpg"
        )
        # you might need to translate '#' links to direct links, e.g.
        direct_url = f"https://apps.usgs.gov/hivis/images/{SITE}___{ts_str}.jpg"
        
        resp = requests.head(url)  # test if it exists
        if resp.status_code == 200:
            return url
    return None

# Example: fetch the closest image to 22:10 UTC
dt = datetime(2025, 8, 1, 0, 10)
print(build_url(dt))
```
