```python
# hivis scraper
# should take a single (most recent) image url from hivis site and sheet with usgs data for that site to obtain image urls.
# The seconds need to be guessed in the timestamp

# CAN:
# - download multiple images based on the timestamp, guesses seconds

# TODO:
# [X] make a loop to get multiple image urls
 # [ ] input excel sheet with usgs data and append urls in new column (work on usgs scraper first) !!master sheet should have cam urls!!
 # OR
 # [ ] call single usgs api call and start the excel sheet here
# [ ] second guesser can be more biased (most seconds are the same)

import requests
from pathlib import Path

# URL of the image you want to download
image_url_full = "https://usgs-nims-images.s3.amazonaws.com/720/NJ_Delaware_River_at_Lambertville_NJ/NJ_Delaware_River_at_Lambertville_NJ___2026-01-28T07-00-08Z.jpg"
image_url_pre = "https://usgs-nims-images.s3.amazonaws.com/720/NJ_Delaware_River_at_Lambertville_NJ/NJ_Delaware_River_at_Lambertville_NJ___2026-01-28T"
image_url_post = "Z.jpg"

print("📌 Starting image download script...")
for hours in range(24):
    for seconds in range(60):

        h = f"{hours:02d}"
        s = f"{seconds:02d}"
        image_url = image_url_pre + h + "-00-" + s + image_url_post
    
        print(f"Trying: {h} {s}")
    
        try:
            # Send the HTTP GET request
            response = requests.get(image_url, stream=True)
            
            # Check response status
            print(f"📡 Server responded with status code: {response.status_code}")
            
            if response.status_code == 200:
                output_filename = "NJ_Delaware_River_at_Lambertville_NJ_2026-01-28_" + h + "-00-" + s + "Z.jpg"
                # Make sure the directory exists for saving
                output_path = Path(output_filename)
                
                # Open a file in write-binary mode
                with open(output_path, "wb") as f:
                    # Loop over chunks of data
                    for chunk in response.iter_content(chunk_size=8192):
                        if chunk:
                            f.write(chunk)
                
                print(f"✅ Download successful! Image saved to: {output_path.resolve()}")
                break
        
        except requests.exceptions.RequestException as e:
            print("⚠️ An error occurred during the download:")
            print(e)
        
print("🏁 Script finished.")
```
