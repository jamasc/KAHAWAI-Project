```python
# TLS Timelapse Frame Extractor
# Converts all .tls files in a folder to .avi and extracts all frames to subfolders using ffmpeg

import os
import shutil
import subprocess
from pathlib import Path

# --- CONFIGURATION ---
input_folder = Path(r"D:/T_LAPSE")  # ← change this to your folder path
output_base = input_folder / "frames"
output_base.mkdir(exist_ok=True)

ffmpeg_path = r"C:\Users\marks\AppData\Local\Microsoft\WinGet\Packages\Gyan.FFmpeg_Microsoft.Winget.Source_8wekyb3d8bbwe\ffmpeg-8.0-full_build\bin\ffmpeg.exe"

print(f"📂 Input folder: {input_folder}")
print(f"📸 Output folder: {output_base}\n")

tls_files = sorted(input_folder.glob("*.tls"))
if not tls_files:
    print("⚠️ No .tls files found. Check your folder path.")
else:
    for tls_path in tls_files:
        base_name = tls_path.stem
        avi_path = tls_path.with_suffix(".avi")

        # Copy to .avi (keeps original .tls intact)
        if not avi_path.exists():
            shutil.copy2(tls_path, avi_path)
            print(f"📄 Copied {tls_path.name} → {avi_path.name}")
        else:
            print(f"➡️ {avi_path.name} already exists, skipping copy.")

        # Create subfolder for frames
        frame_folder = output_base / base_name
        frame_folder.mkdir(exist_ok=True)

        # ffmpeg command to extract all frames
        frame_pattern = str(frame_folder / "frame_%04d.jpg")
        cmd = [ffmpeg_path, "-i", str(avi_path), "-q:v", "2", frame_pattern]

        print(f"🎞️ Extracting frames from {avi_path.name} → {frame_folder.name}")
        subprocess.run(cmd, stdout=subprocess.DEVNULL, stderr=subprocess.STDOUT)

    print("\n✅ All TLS files processed and frames extracted successfully!")
```
