```python
# this code takes a datasheet as input and displays information about the data
# plot gage readings over timestamps


import pandas as pd
import plotly.express as px

# Load your data
df = pd.read_excel("pukele_datasheet.xlsx")  # replace with your file path

# Combine date and time into a single datetime column
df['datetime'] = pd.to_datetime(df['date'] + ' ' + df['time'])

# -----------------------------
# 1. Scatter plot: discharge over time, hover shows path
# -----------------------------
fig_scatter = px.scatter(
    df,
    x='datetime',
    y='discharge_cfs',
    hover_data=['path', 'gage_height_ft'],  # info on hover
    title='Discharge over Time'
)
fig_scatter.update_traces(marker=dict(size=6, color='blue', opacity=0.7))
fig_scatter.show()

# -----------------------------
# 2. Histogram: distribution of discharge
# -----------------------------
fig_hist = px.histogram(
    df,
    x='discharge_cfs',
    nbins=30,
    title='Distribution of Discharge',
    hover_data=['path']  # optional, shows path on hover
)
fig_hist.update_traces(marker_color='green', opacity=0.7)
fig_hist.show()

# -----------------------------
# 3. Box plot: summary of discharge
# -----------------------------
fig_box = px.box(
    df,
    y='discharge_cfs',
    points='all',  # shows all points overlaid on box
    hover_data=['path', 'datetime'],  # show path and datetime on hover
    title='Boxplot of Discharge'
)
fig_box.update_traces(marker_color='orange', jitter=0.5)
fig_box.show()
```
