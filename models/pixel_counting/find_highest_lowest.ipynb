```python
import pandas as pd

def find_min_max_row(file_path, sheet_name, column_name):
    # Load the Excel sheet
    df = pd.read_excel(file_path, sheet_name=sheet_name)
    
    # Ensure the column exists
    if column_name not in df.columns:
        print(df.columns)
        raise ValueError(f"Column '{column_name}' not found in sheet '{sheet_name}'")
    
    # Find index of min and max
    idx_min = df[column_name].idxmin()
    idx_max = df[column_name].idxmax()
    
    # Extract rows
    row_min = df.loc[idx_min]
    row_max = df.loc[idx_max]
    
    e_min = row_min['Image_Path']
    e_max = row_max['Image_Path']
    
    return row_min, row_max, e_min, e_max

# Example usage:
file_path = "C:/Users/marks/pythonscripts/pixel_counting/image_inventory_cam_1000.xlsx"
sheet_names = ["Label_1", "Label_2", "Label_3", "Label_4", "Label_5", "Label_6"]
column_name = "water"

for sheet_name in sheet_names:
    print(sheet_name)
    row_min, row_max, e_min, e_max= find_min_max_row(file_path, sheet_name, column_name)
    print("min path", e_min)
    print("max path", e_max)
    print('##################################################################')
```
