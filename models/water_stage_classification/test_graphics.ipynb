```python
# notbook to use test1.py and generate confusion matrices
# run this cell first

import test1
import importlib
import pandas as pd
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt
import yaml

#importlib.reload(test1)
```

```python
# this code creates a confusion matrix for the accurracy of predictions
# for individual DAR

all_y_true = []
all_y_pred = []

for i in range(7):

    path = f"config_DAR100{i}.yaml"

    results = test1.main(path)

    y_true = [t.item() + 1 for t in results["label"]]
    y_pred = [t.item() + 1 for t in results["predicted"]]

    all_y_true.extend(y_true)
    all_y_pred.extend(y_pred)

    # Compute confusion matrix
    labels = [2, 3, 1, 4]
    display_labels = ['dry bed', 'isolated pools', 'run present', 'freshet']
    cm = confusion_matrix(y_true, y_pred, labels=labels)
    
    print(f"Confusion Matrix (DAR100{i}):")
    print(cm)
    
    # Plot it
    disp = ConfusionMatrixDisplay(
        confusion_matrix=cm,
        display_labels=display_labels
    )
    disp.plot(cmap="Blues")
    plt.title("Confusion Matrix")
    plt.show()

labels = [2, 3, 1, 4]
display_labels = ['dry bed', 'isolated pools', 'run present', 'freshet']
cm = confusion_matrix(all_y_true, all_y_pred, labels=labels)

print(f"Confusion Matrix (combined):")
print(cm)

# Plot it
disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=display_labels
)
disp.plot(cmap="Blues")
plt.title("Confusion Matrix")
plt.savefig("confusion_matrix.jpg", dpi=300, bbox_inches="tight")
plt.show()
```
