```python
# notbook to use test1.py and generate confusion matrices

import test1
import pandas as pd
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt


#test_acc, test_loss, label_counts, label_correct, label_confidences = test1.main('config_DAR1000.yaml')
results = test1.main('config_DAR1000.yaml')
```

```python
# this code creates a confusion matrix for the accurracy of predictions

# Replace with your actual column names
y_true = results["label"]
y_pred = results["predicted"]

# Compute confusion matrix
labels = [2, 3, 1, 4]
display_labels = ['dry bed', 'isolated pools', 'run present', 'freshet']
cm = confusion_matrix(y_true, y_pred, labels=labels)

print("Confusion Matrix:")
print(cm)

# Plot it
disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=display_labels
)
disp.plot(cmap="Blues")
plt.title("Confusion Matrix")
plt.show()
```
